# Experiment 5: Geo-Visualization with Maps

Aggregate sales by state and display an interactive symbol map using sample coordinates.

```python
import pandas as pd
import plotly.express as px
df = pd.read_csv("data/retail_sales.csv")
state_sales = df.groupby("State",as_index=False)["Sales"].sum()
coords = pd.DataFrame({
 "State":["Delhi","Karnataka","West Bengal","Maharashtra","Tamil Nadu"],
 "Latitude":[28.6139,15.3173,22.9868,19.7515,11.1271],
 "Longitude":[77.2090,75.7139,87.8550,75.7139,78.6569]})
map_df = state_sales.merge(coords,on="State",how="left")
display(map_df)
```

## Symbol map

```python
fig = px.scatter_geo(map_df,lat="Latitude",lon="Longitude",size="Sales",color="Sales",
                     hover_name="State",title="State-wise Retail Sales",projection="natural earth")
fig.show()
fig.write_html("05_Geo_Visualization.html",include_plotlyjs=True)
```
