# Experiment 8: Time Series Forecasting and Trend Lines

Calculate moving averages, fit a linear trend, and forecast three future monthly periods.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
df = pd.read_csv("data/retail_sales.csv",parse_dates=["OrderDate"])
monthly = df.set_index("OrderDate").resample("MS")["Sales"].sum().reset_index()
monthly["Time"] = np.arange(len(monthly))
display(monthly)
```

## Moving average and trend line

```python
monthly["Moving_Average"] = monthly.Sales.rolling(3,min_periods=1).mean()
model = LinearRegression().fit(monthly[["Time"]],monthly["Sales"])
monthly["Trend"] = model.predict(monthly[["Time"]])
plt.figure(figsize=(9,5))
plt.plot(monthly.OrderDate,monthly.Sales,marker="o",label="Actual Sales")
plt.plot(monthly.OrderDate,monthly.Moving_Average,label="3-month Moving Average")
plt.plot(monthly.OrderDate,monthly.Trend,linestyle="--",label="Linear Trend")
plt.title("Monthly Sales Trend"); plt.xlabel("Month"); plt.ylabel("Sales")
plt.legend(); plt.grid(True); plt.tight_layout(); plt.show()
```

## Forecast next three months

```python
future_x = np.arange(len(monthly),len(monthly)+3).reshape(-1,1)
future_y = model.predict(future_x)
future_dates = pd.date_range(monthly.OrderDate.max()+pd.offsets.MonthBegin(1),periods=3,freq="MS")
forecast = pd.DataFrame({"Month":future_dates,"Forecast_Sales":future_y})
display(forecast)
```
