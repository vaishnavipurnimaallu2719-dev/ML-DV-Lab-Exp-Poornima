# Experiment 4: Interactive Dashboard

Create interactive Plotly charts with hover details and export an HTML dashboard.

```python
import pandas as pd
import plotly.express as px
df = pd.read_csv("data/retail_sales.csv", parse_dates=["OrderDate"])
```

## Interactive bar chart

```python
fig = px.bar(df, x="Category", y="Sales", color="Region",
             hover_data=["OrderDate","Subcategory","Profit"],
             title="Retail Sales by Category and Region")
fig.show()
```

## Interactive line chart

```python
daily = df.groupby("OrderDate", as_index=False)["Sales"].sum()
fig = px.line(daily, x="OrderDate", y="Sales", markers=True, title="Sales Trend")
fig.show()
```

## Export an HTML dashboard

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
fig = make_subplots(rows=1, cols=2, subplot_titles=("Sales by Category","Sales Trend"))
cat = df.groupby("Category",as_index=False)["Sales"].sum()
fig.add_trace(go.Bar(x=cat.Category,y=cat.Sales,name="Category Sales"),row=1,col=1)
trend = df.groupby("OrderDate",as_index=False)["Sales"].sum()
fig.add_trace(go.Scatter(x=trend.OrderDate,y=trend.Sales,mode="lines+markers",name="Daily Sales"),row=1,col=2)
fig.update_layout(title="Interactive Retail Sales Dashboard",height=500)
fig.write_html("04_Interactive_Dashboard.html",include_plotlyjs=True)
print("Created 04_Interactive_Dashboard.html")
```
