# Experiment 6: Pie and Donut Charts

Show proportions using an Olympic-medal-style sample dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
medals = pd.read_csv("data/olympic_medals.csv")
display(medals)
```

## Pie chart

```python
plt.figure(figsize=(7,7))
plt.pie(medals.Medals,labels=medals.Country,autopct="%1.1f%%",startangle=90)
plt.title("Medal Share by Country - Pie Chart"); plt.show()
```

## Donut chart

```python
plt.figure(figsize=(7,7))
plt.pie(medals.Medals,labels=medals.Country,autopct="%1.1f%%",startangle=90,
        wedgeprops={"width":0.42})
plt.title("Medal Share by Country - Donut Chart"); plt.show()
```
