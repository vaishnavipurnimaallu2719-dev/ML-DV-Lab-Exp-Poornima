# Experiment 2: Data Cleaning and Transformation

Handle missing values, rename columns, convert data types, and create calculated fields.

```python
import pandas as pd
df = pd.read_csv("data/retail_sales.csv")
df.loc[1,"Profit"] = None
df.loc[3,"Category"] = None
df.loc[5,"Quantity"] = None
print("Missing values before cleaning:"); display(df.isna().sum())
```

## Clean and transform

```python
df.rename(columns={"OrderDate":"Order_Date","Sales":"Sales_Amount"}, inplace=True)
df["Category"] = df["Category"].fillna("Unspecified")
df["Profit"] = df["Profit"].fillna(df["Profit"].median())
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
df["Profit_Margin"] = df["Profit"] / df["Sales_Amount"]
display(df.head()); display(df.isna().sum())
```

## Save cleaned CSV

```python
df.to_csv("data/retail_sales_cleaned.csv", index=False)
print("Saved data/retail_sales_cleaned.csv")
```
