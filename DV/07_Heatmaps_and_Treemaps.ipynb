# Experiment 7: Heatmaps and Tree Maps

Create a correlation heatmap and a category/subcategory sales treemap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
df = pd.read_csv("data/retail_sales.csv")
```

## Correlation heatmap

```python
corr = df[["Sales","Profit","Quantity","Discount"]].corr()
plt.figure(figsize=(7,5))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Retail Data Correlation Heatmap"); plt.tight_layout(); plt.show()
```

## Treemap

```python
tree = df.groupby(["Category","Subcategory"],as_index=False)["Sales"].sum()
fig = px.treemap(tree,path=["Category","Subcategory"],values="Sales",
                 title="Sales Treemap by Category and Subcategory")
fig.show()
fig.write_html("07_Heatmaps_and_Treemaps.html",include_plotlyjs=True)
```
