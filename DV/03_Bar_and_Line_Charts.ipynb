# Experiment 3: Bar and Line Charts

Compare categories with a bar chart and visualize trends over time with a line chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("data/retail_sales.csv", parse_dates=["OrderDate"])
```

## Bar chart with labels

```python
ax = df.groupby("Category")["Sales"].sum().plot(kind="bar", figsize=(8,5), title="Sales by Category")
ax.set_xlabel("Category"); ax.set_ylabel("Sales")
for c in ax.containers: ax.bar_label(c)
plt.tight_layout(); plt.show()
```

## Line chart with labels

```python
monthly = df.set_index("OrderDate").resample("MS")["Sales"].sum()
ax = monthly.plot(marker="o", figsize=(8,5), title="Monthly Sales Trend")
ax.set_xlabel("Month"); ax.set_ylabel("Sales"); ax.grid(True)
for x,y in zip(monthly.index, monthly.values):
    ax.annotate(str(int(y)), (x,y), textcoords="offset points", xytext=(0,8), ha="center")
plt.tight_layout(); plt.show()
```
